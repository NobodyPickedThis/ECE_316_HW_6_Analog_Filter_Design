### 5. Based on your pole-zero plots from item 4, identify which filter families produce zeros exactly on the imaginary axis of the s-plane. Explain what a zero at $s = ±jω_0$ implies for the magnitude $|H(jω_0)|$ and for the phase response as ω passes through $ω_0$.

Chebyshev and Elliptic filters produce zeros on the imaginary axis. A zero on this axis implies a magnitude of zero at the specified $\omega$ value, and will produce a 180° phase jump at that point.
