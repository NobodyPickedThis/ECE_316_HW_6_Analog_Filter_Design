### 12. Written explanation of the numerical failure mechanism and how to recognize it.

As poles get closer to the imaginary axis (which happens with increased order filters), the system will become unstable with smaller inaccuracies than a lower-order equivalent (think component tolerances). In this case, our tolerance is the accuracy of the floating point representation we're using. More accurate types could be used (e.g. doubles), but this appears to be less common due to the tradeoffs involved, such as increased computation requirements, being less worth it.  
  
The most obvious symptom is that the filter does not meet the specification it was generated to match. Furthermore, the Butterworth filter family is monotonically decreasing, but we see clear oscillation in the graph. Another possible symptom is discontinuity, but we don't see any here. Regardless, if a graph takes on a shape its function should not be able to generate, it's likely unstable.
